## 4. Feature Selection

We use **Annual Income** and **Spending Score** for customer segmentation.

```python
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print('Feature matrix shape:', X.shape)
```

## 5. Elbow Method

```python
wcss = []

for k in range(1, 11):
    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(1,11), wcss, marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.xticks(range(1,11))
plt.show()
```

## 6. K-Means Clustering with K = 4

```python
kmeans = KMeans(
    n_clusters=4,
    init='k-means++',
    random_state=42,
    n_init=10
)

df['Cluster'] = kmeans.fit_predict(X_scaled)
df.head(10)
```

```python
plt.figure(figsize=(15,8))

sns.scatterplot(
    data=df,
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Cluster',
    palette='viridis',
    edgecolors='black',
    s=120
)

centers = scaler.inverse_transform(kmeans.cluster_centers_)
plt.scatter(
    centers[:, 0],
    centers[:, 1],
    s=200,
    marker='X',
    c='red',
    label='Centroids'
)

plt.title('Customer Segmentation using K-Means Clustering')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()
```

## 7. Cluster Summary

```python
cluster_summary = df.groupby('Cluster').agg(
    Customers=('CustomerID', 'count'),
    Average_Age=('Age', 'mean'),
    Average_Income=('Annual Income (k$)', 'mean'),
    Average_Spending=('Spending Score (1-100)', 'mean')
).round()

cluster_summary
```

## 8. Silhouette Score

```python
silhouette = silhouette_score(X_scaled, df['Cluster'])
print('Silhouette Score:', round(silhouette, 2))
```

## 10. Conclusion

### Business Insights from Customer Segmentation

The K-Means clustering algorithm identified four distinct customer segments based on their income and spending behavior. These segments provide useful insights for developing targeted marketing strategies, improving customer engagement, and increasing sales.

**Cluster 0 – Low-Income, Low-Spending Customers:** This cluster contains 23 customers with an average age of 45 years, an average annual income of approximately $26k, and an average spending score of 21. These customers have relatively low incomes and low spending activity. Businesses can target this segment with affordable products, discounts, and budget-friendly offers.

**Cluster 1 – High-Income, Low-Spending Customers:** This cluster consists of 38 customers with an average age of 40 years, an average annual income of approximately $87k, and an average spending score of 19. Although these customers earn relatively high incomes, their spending is low. Businesses can explore personalized promotions, loyalty programs, and premium product recommendations to encourage greater engagement and purchasing activity.

**Cluster 2 – Moderate-Income, Moderate-Spending Customers:** This is the largest cluster, containing 100 customers, with an average age of 39 years, an average annual income of approximately $48k, and an average spending score of 56. These customers represent half of the dataset and demonstrate moderate spending behavior. Businesses can target them with product recommendations, cross-selling opportunities, seasonal offers, and general marketing campaigns.

**Cluster 3 – High-Income, High-Spending Customers:** This cluster contains 39 customers with an average age of 33 years, an average annual income of approximately $87k, and an average spending score of 82. These relatively younger customers have high incomes and high spending scores. Businesses should focus on customer retention, premium products, exclusive offers, and personalized shopping experiences to maintain their engagement and encourage repeat purchases.

**Overall Conclusion:** The analysis demonstrates that customers have different income levels and spending patterns and should not be treated as a single homogeneous group. By tailoring marketing campaigns and product offerings to each segment, businesses can allocate promotional resources more effectively, improve customer satisfaction, identify potential growth opportunities, and build stronger customer relationships.
